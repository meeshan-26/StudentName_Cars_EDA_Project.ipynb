```python
import zipfile
import pandas as pd

zip_path = "/content/MohamedMeeshan_Cars_EDA_Project.ipynb.ipynb"

# Open the zip archive and read the CSV file inside it
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    file_list = zip_ref.namelist()
    print("Files inside archive:", file_list)

    # Find the CSV file inside the zip archive
    csv_file = [f for f in file_list if f.endswith('.csv')][0]
    with zip_ref.open(csv_file) as f:
        df = pd.read_csv(f)

display(df.head())
```

```python
# Univariate Analysis
sns.distplot(df['Warranty Service Km'])
plt.show()
```

```python
#Outlier Treatment
sns.boxplot(df[['Identification Body Type','Warranty Service Km']])
plt.title("Outlier detection")
plt.show()
```

```python
#Outlier treatment
# Only apply IQR outlier capping to numerical columns
for col in ['Warranty Service Km']:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1

  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR

  df[col] = np.where(df[col] > upper, upper, np.where(df[col] < lower, lower, df[col]))
```

```python
#Outlier Treatment
sns.boxplot(df[['Warranty Service Km']])
plt.title("Outlier detection")
plt.show()
```
